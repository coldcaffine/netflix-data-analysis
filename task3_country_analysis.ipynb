```python
import pandas as pd
df = pd.read_csv("/Users/aditivdharme/Desktop/netflix-data-analysis/netflix_cleaned.csv")
df.head()
```

```python
# split "India, United States" into a list, then give each country its own row
countries = df["country"].str.split(",").explode().str.strip()

# remove the "Unknown" rows we filled in during cleaning
countries = countries[countries != "Unknown"]

# count titles per country
country_counts = countries.value_counts()

# show the top 10
country_counts.head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

top10 = country_counts.head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top10.values, y=top10.index, color="steelblue")
plt.title("Top 10 Countries by Number of Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()
```
